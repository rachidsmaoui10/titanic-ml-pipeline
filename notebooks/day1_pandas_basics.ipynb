```python
X = df[["Sex", "Pclass", "Age"]]
y = df["Survived"]
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
import pandas as pd

coef = pd.Series(
    model.coef_[0],
    index=X_encoded.columns
).sort_values(ascending=False)

coef
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X_train, y_train)
```

```python
y_pred_tree = tree.predict(X_test)

from sklearn.metrics import accuracy_score
accuracy_tree = accuracy_score(y_test, y_pred_tree)
accuracy_tree
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred_tree)
```

```python
train_acc_log = accuracy_score(y_train, model.predict(X_train))
test_acc_log = accuracy_score(y_test, model.predict(X_test))

train_acc_tree = accuracy_score(y_train, tree.predict(X_train))
test_acc_tree = accuracy_score(y_test, tree.predict(X_test))

train_acc_log, test_acc_log, train_acc_tree, test_acc_tree
```

```python
df["IsChild"]  = (df["Age"] < 15).astype(int)
df["IsMother"] = ((df["Sex"] == "female") & (df["Parch"] > 0)).astype(int)
df["IsAlone"]  = ((df["SibSp"] == 0) & (df["Parch"] == 0)).astype(int)
```

```python
X2 = df[["Sex", "Pclass", "Age", "IsChild", "IsMother", "IsAlone"]]
y  = df["Survived"]
```

```python
X2_enc = pd.get_dummies(X2, drop_first=True)
```

```python
X2_train, X2_test, y_train, y_test = train_test_split(
    X2_enc, y, test_size=0.2, random_state=42
)
```

```python
model2 = LogisticRegression(max_iter=1000)
model2.fit(X2_train, y_train)
```

```python
accuracy_score(y_test, model2.predict(X2_test))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

y2_pred = model2.predict(X2_test)

confusion_matrix(y_test, y2_pred)
```

```python
print(classification_report(y_test, y2_pred))
```

```python
y2_prob = model2.predict_proba(X2_test)[:,1]
```

```python
import numpy as np
from sklearn.metrics import recall_score

for t in [0.3, 0.4, 0.5, 0.6]:
    y_pred_t = (y2_prob >= t).astype(int)
    print("Seuil", t, "Recall survivants =", recall_score(y_test, y_pred_t))
```

```python
y_pred_03 = (y2_prob >= 0.3).astype(int)
confusion_matrix(y_test, y_pred_03)
```
